```python
# ==============================================================================
# TỰ ĐỘNG NÉN TOÀN BỘ KẾT QUẢ VÀ TẢI VỀ MÁY TÍNH (GOOGLE COLAB)
# ==============================================================================
import os
import shutil
from google.colab import files

print('=' * 70)
print('ĐANG NÉN TOÀN BỘ KẾT QUẢ (CHECKPOINTS, LOGS, JSON)...')
print('=' * 70)

if os.path.exists('outputs'):
    zip_filename = 'segformer_all_outputs'
    zip_path = f'{zip_filename}.zip'
    
    # Xóa file zip cũ nếu có từ lần chạy trước
    if os.path.exists(zip_path):
        os.remove(zip_path)
        
    # Nén thư mục outputs
    shutil.make_archive(zip_filename, 'zip', 'outputs')
    size_mb = os.path.getsize(zip_path) / (1024 * 1024)
    print(f'✓ Đã nén thành công file: {zip_path} ({size_mb:.2f} MB)')
    print('✓ Đang kích hoạt tải file về máy qua trình duyệt...')
    
    # Tải file về máy
    files.download(zip_path)
else:
    print('❌ Lỗi: Không tìm thấy thư mục outputs!')
```
